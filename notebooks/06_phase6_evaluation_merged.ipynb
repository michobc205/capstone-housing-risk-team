# Phase 6 — Evaluation on the Test Period

**Purpose.** Score the nine models of Phase 5 on the Test period with frozen models and settings, and answer three questions: how good are the scores on a later period, does the mortgage rate add anything on Test (where the 2022-23 rate-hiking cycle sits), and how should the predictions be used (decision rules, calibration, stability over time).

**How Test is treated here.** Every choice (the families, the settings, the feature steps, the recommended model and the decision cut-offs) was made on the validation folds. Test is scored with frozen models. Test was also scored in earlier, retired versions of both independent pipelines, so it is not a fresh draw. The protection is that nothing in this notebook is chosen from Test results: if a Test result disagrees with a validation result, the validation-based choice stands and the difference is reported.

## Design, fixed before any result

| Item | Choice |
|---|---|
| Models | All 9 of Phase 5 (3 families x 3 feature steps), same fixed settings, refit on **all** Train rows |
| Test scoring | With frozen models. The scores are locked in `docs/phase6_test_lock.json`. A re-run must reproduce them (per-family tolerance) or the notebook stops |
| Metrics | ROC-AUC (main), PR-AUC and Brier score, read against the Test base rate (8.11%) |
| Does the rate help on Test? | Same rule as Phase 5: the interval of the change excludes zero **and** the gain is at least +0.01 ROC-AUC. A step is robust if it helps in at least 2 of the 3 families. Bootstrap over the 17 Test quarters, 2,000 draws |
| Decision rules | Flag the riskiest 5%, 10% and 20% of counties **in each quarter**, and flag every county whose probability is at or above the F1-maximizing cut-off found on the validation predictions. No rule is picked: all four are reported side by side, and the cut-off rule is judged by lift |
| Reference | The same rules computed on the saved validation predictions, so Test can be compared with validation |
| Calibration | Reported on Test and on validation, with no recalibration |
| Stability | ROC-AUC, PR-AUC and recall at the top 10% for each of the 17 Test quarters |

**What this notebook does not do.** It does not tune anything, does not pick a different model because of Test, and does not recalibrate.

## 4. Decision rules chosen on validation, before Test is scored

Phase 5 saved the out-of-sample predictions of every model on the 11 validation folds. From them, **before Test predictions exist**, this section finds the probability cut-off that maximizes F1 (one cut-off per model, from all validation rows pooled) and computes the reference results of every decision rule.

The four rules are:

- **Top 5%, 10%, 20% in each quarter:** in every quarter, flag the counties with the highest predicted probability, as many as that share of the quarter's counties. The rule depends only on the ranking within a quarter, so it does not depend on the base rate.
- **F1 cut-off:** flag every county whose probability is at or above the cut-off found here on validation, applied unchanged to Test. It does depend on the base rate, and Test has more declines than validation, so it will flag a different share there.

**Port checks (addition from the team merge).** Before any cut-off is computed, the validation predictions are checked: 11 folds, 44 quarters, 112,734 rows and 4,252 declines, at least 20 declines in every fold, no duplicate county and quarter, no missing or impossible probability, every row inside Train with the same label as in the modeling table, and all validation labels end before the first Test label quarter starts.

**Cut-offs are frozen.** The first run computes the cut-offs, checks them against the reference cut-offs of the independent pipeline and writes `docs/phase6_thresholds_lock.json`. Every later run, on any machine, recomputes them only as a check and then **applies the cut-offs of the committed file**, so the rule is the same everywhere. The check uses the per-family tolerance: logistic regression and histogram gradient boosting must reproduce all four numbers (cut-off, precision, recall, F1) to within 1e-6. For the random forest two things must hold, each to within **1e-2** (team decision, 2026-10-07): (1) the best validation F1 matches the locked F1, and (2) the locked cut-off, applied to this machine's validation predictions, gives an F1 within 1e-2 of this machine's best F1. The tolerance is wider than the 1e-3 used for the Phase 5 baseline because the F1 curve of a random forest is flat near its top: a small difference between machines can move the best cut-off by a few thousandths while the F1, which is about 0.44, hardly changes. Everything that would reveal a real problem stays strict (the other two families at 1e-6, their Test scores, the row counts). The gap between the cut-off values is printed for information and is never a reason to stop on its own. The same second check (the locked cut-off applied to this machine's predictions) is run for every model.

**Feature steps are recorded.** The lock file also lists the columns of all three feature steps. The notebook stops unless they match the fixed lists and the momentum columns of the committed Phase 5 baseline lock (which is not touched). This only records the features before Test is scored and changes no model.

```python
def top_share_counts(y, p, q, share):
    # Flag the top `share` of rows inside each quarter; return TP, FP, FN, TN summed over quarters.
    tp = fp = fn = tn = 0
    for qv in np.unique(q):
        m = q == qv
        yq, pq = y[m], p[m]
        k = max(1, int(round(share * len(yq))))
        flag = np.zeros(len(yq), dtype=bool)
        flag[np.argsort(-pq, kind="mergesort")[:k]] = True
        tp += int((flag & (yq == 1)).sum()); fp += int((flag & (yq == 0)).sum())
        fn += int((~flag & (yq == 1)).sum()); tn += int((~flag & (yq == 0)).sum())
    return tp, fp, fn, tn


def threshold_counts(y, p, t):
    flag = p >= t
    return (int((flag & (y == 1)).sum()), int((flag & (y == 0)).sum()),
            int((~flag & (y == 1)).sum()), int((~flag & (y == 0)).sum()))


def rule_row(split, model, rule, counts):
    tp, fp, fn, tn = counts
    n = tp + fp + fn + tn
    prec = tp / (tp + fp) if tp + fp else float("nan")
    rec_ = tp / (tp + fn) if tp + fn else float("nan")
    base = (tp + fn) / n
    return {"split": split, "family": model[0], "step": model[1], "rule": rule,
            "tp": tp, "fp": fp, "fn": fn, "tn": tn, "precision": prec, "recall": rec_,
            "flagged_share": (tp + fp) / n, "lift": prec / base if base else float("nan")}


RULE_NAMES = [f"top {int(100 * s)}% each quarter" for s in TOP_SHARES] + ["F1 threshold from validation"]
```

```python
# Reference cut-offs: copied from the independent SQL pipeline's phase6_thresholds_lock.json.
REFERENCE_THRESHOLDS = {
    "logreg|baseline": {"threshold": 0.28050005573505504, "val_precision": 0.41033672670321064,
                        "val_recall": 0.4929444967074318, "val_f1": 0.447863247862752},
    "logreg|step2_rate_level": {"threshold": 0.2572964902292658, "val_precision": 0.4177984998986418,
                                "val_recall": 0.48471307619943554, "val_f1": 0.44877517691839225},
    "logreg|step3_rate_change": {"threshold": 0.23791861749021523, "val_precision": 0.398300283286119,
                                 "val_recall": 0.4960018814675447, "val_f1": 0.441814182465202},
    "random_forest|baseline": {"threshold": 0.3385948219428074, "val_precision": 0.40388685780257844,
                               "val_recall": 0.49365004703668863, "val_f1": 0.44427981796966065},
    "random_forest|step2_rate_level": {"threshold": 0.33266557248650896, "val_precision": 0.3801923076923077,
                                       "val_recall": 0.4649576669802446, "val_f1": 0.41832416419755836},
    "random_forest|step3_rate_change": {"threshold": 0.33085225251509903, "val_precision": 0.39555369127516776,
                                        "val_recall": 0.4435559736594544, "val_f1": 0.41818181818131983},
    "hist_gb|baseline": {"threshold": 0.3894659892931262, "val_precision": 0.4359499431171786,
                         "val_recall": 0.4506114769520226, "val_f1": 0.44315947727485583},
    "hist_gb|step2_rate_level": {"threshold": 0.27835343894122355, "val_precision": 0.3285289747399703,
                                 "val_recall": 0.5199905926622765, "val_f1": 0.4026588963754133},
    "hist_gb|step3_rate_change": {"threshold": 0.27534866951140213, "val_precision": 0.3314135264110581,
                                  "val_recall": 0.47365945437441204, "val_f1": 0.38996998741359246},
}


# Team decision (2026-10-07): for the random forest the two F1 checks use 1e-2, not 1e-3. Near its maximum the F1
# curve of a random forest is flat, and an F1 of about 0.44 changes little within 0.01. Logistic regression and
# histogram gradient boosting stay at 1e-6 on all four numbers.
RF_F1_TOLERANCE = 1e-2
F1_AT_CUTOFF_TOLERANCE = {"logreg": 1e-3, "hist_gb": 1e-3, "random_forest": RF_F1_TOLERANCE}


def compare_cutoffs(new, old):
    # Per-family rule. Returns (problems, notes); `problems` stop the notebook, `notes` are only printed.
    problems, notes = [], []
    for key, ref_vals in old.items():
        fam = key.split("|")[0]
        tol = REPRODUCTION_TOLERANCE[fam]
        gaps = {k: abs(float(new[key][k]) - float(ref_vals[k])) for k in ref_vals}
        if fam == "random_forest":
            if gaps["val_f1"] >= RF_F1_TOLERANCE:
                problems.append(f"{key}: best validation F1 differs by {gaps['val_f1']:.2e} (tolerance {RF_F1_TOLERANCE:.0e})")
            elif gaps["threshold"] >= tol:
                notes.append(f"{key}: the cut-off moved by {gaps['threshold']:.2e} but the best validation F1 differs "
                             f"by only {gaps['val_f1']:.2e}; the frozen cut-off is applied, nothing stops")
        else:
            worst = max(gaps, key=gaps.get)
            if gaps[worst] >= tol:
                problems.append(f"{key}: {worst} differs by {gaps[worst]:.2e} (tolerance {tol:.0e})")
    return problems, notes


def f1_at_cutoff(y, p, t):
    flag = p >= t
    tp = int((flag & (y == 1)).sum())
    fp = int((flag & (y == 0)).sum())
    fn = int((~flag & (y == 1)).sum())
    return 2 * tp / (2 * tp + fp + fn) if (2 * tp + fp + fn) else 0.0


def cutoff_report(new, anchor):
    # Two checks plus information. (1) compare_cutoffs: per-family rule on the numbers.
    # (2) The anchor cut-off, applied to THIS machine's validation predictions, must give an F1 within 1e-3 of
    #     this machine's best F1, so applying the frozen cut-off is harmless on every machine.
    # The gap between the cut-off values is printed for information and never stops the notebook by itself.
    problems, notes = compare_cutoffs(new, anchor)
    rows = []
    for key, vals in anchor.items():
        m = tuple(key.split("|"))
        gap_here = abs(f1_at_cutoff(val_y, val_p[m], vals["threshold"]) - new[key]["val_f1"])
        rows.append({"model": key, "cutoff_gap": abs(new[key]["threshold"] - vals["threshold"]),
                     "best_f1_gap": abs(new[key]["val_f1"] - vals["val_f1"]),
                     "f1_gap_with_anchor_cutoff": gap_here})
        tol_here = F1_AT_CUTOFF_TOLERANCE[m[0]]
        if gap_here >= tol_here:
            problems.append(f"{key}: the anchor cut-off gives an F1 that is {gap_here:.2e} below this machine's best "
                            f"(tolerance {tol_here:.0e})")
    return problems, notes, pd.DataFrame(rows)


f1_threshold_new = {}
for m in MODELS:
    prec, rec_, thr = precision_recall_curve(val_y, val_p[m])
    f1 = 2 * prec * rec_ / (prec + rec_ + 1e-12)
    i = int(np.nanargmax(f1[:-1]))
    f1_threshold_new[f"{m[0]}|{m[1]}"] = {"threshold": float(thr[i]), "val_precision": float(prec[i]),
                                          "val_recall": float(rec_[i]), "val_f1": float(f1[i])}

def show_report(report):
    print("Cut-off check (information; the cut-off gap alone never stops the notebook):")
    print(report.to_string(index=False, float_format=lambda v: f"{v:.2e}"))


if THR_LOCK.exists():
    old = json.loads(THR_LOCK.read_text(encoding="utf-8"))
    assert (old["validation_rows"], old["validation_declines"]) == (len(vp), int(val_y.sum())), (
        "STOP: the thresholds lock was written for different validation data.")
    assert old.get("feature_steps") == FEATURE_STEPS, "STOP: the feature steps in the thresholds lock differ from this notebook."
    problems, notes, report = cutoff_report(f1_threshold_new, old["thresholds"])
    ref_problems, _ = compare_cutoffs(old["thresholds"], REFERENCE_THRESHOLDS)
    assert not ref_problems, "STOP: the locked cut-offs do not match the reference cut-offs: " + "; ".join(ref_problems)
    show_report(report)
    assert not problems, ("STOP: the recomputed cut-offs no longer match the locked file: " + "; ".join(problems)
                          + f". Library versions in the lock: {old.get('library_versions')}; now: {LIB_VERSIONS}. "
                          f"Delete {THR_LOCK} only if the validation predictions changed on purpose.")
    for n in notes:
        print("NOTE:", n)
    THR_USED = old["thresholds"]
    print("The recomputed cut-offs reproduce the locked file. The locked (frozen) cut-offs are applied.")
else:
    problems, notes, report = cutoff_report(f1_threshold_new, REFERENCE_THRESHOLDS)
    show_report(report)
    assert not problems, ("STOP: the cut-offs do not reproduce the reference cut-offs of the independent pipeline: "
                          + "; ".join(problems) + ". Nothing was locked.")
    for n in notes:
        print("NOTE:", n)
    thr_lock = {"purpose": "F1 cut-offs found on the Phase 5 validation predictions, written before Test was scored. "
                           "A re-run applies these cut-offs unchanged.",
                "validation_rows": int(len(vp)), "validation_declines": int(val_y.sum()),
                "library_versions": LIB_VERSIONS, "feature_steps": FEATURE_STEPS,
                "thresholds": f1_threshold_new}
    THR_LOCK.parent.mkdir(parents=True, exist_ok=True)
    THR_LOCK.write_text(json.dumps(thr_lock, indent=2), encoding="utf-8")
    THR_USED = f1_threshold_new
    print(f"Cut-offs reproduce the reference and are locked in {THR_LOCK}. Commit it before Test is scored (the next cell checks this).")

f1_threshold = {tuple(k.split("|")): v for k, v in THR_USED.items()}
```

## 7. Does the rate help on Test? Bootstrap over the 17 Test quarters

The same method as Phase 5, with the Test quarters as the unit: the 17 quarters are resampled with replacement 2,000 times, and every draw scores each model on the quarters drawn. A step "helps" only if its interval excludes zero **and** the change is at least +0.01 ROC-AUC. The same draws give the interval for every model's ROC-AUC and for the differences between families. The first cell proves the weighted AUC equals scikit-learn.

```python
def weighted_auc(gid, yy, w):
    n_g = gid[-1] + 1
    wp = np.bincount(gid, weights=w * yy, minlength=n_g)
    wn = np.bincount(gid, weights=w * (1.0 - yy), minlength=n_g)
    P, N = wp.sum(), wn.sum()
    if P == 0 or N == 0:
        return np.nan
    neg_before = np.cumsum(wn) - wn
    return float((wp * (neg_before + 0.5 * wn)).sum() / (P * N))


qid = {q: i for i, q in enumerate(test_quarters)}
test_qidx = test["quarter"].map(qid).to_numpy()
n_q = len(qid)

prep = {}
for m in MODELS:
    order = np.argsort(test_p[m], kind="mergesort")
    ss = test_p[m][order]
    gid = np.cumsum(np.r_[True, ss[1:] != ss[:-1]]) - 1
    prep[m] = (gid, test_y[order].astype(float), test_qidx[order])

max_gap = max(abs(weighted_auc(g, yy, np.ones(len(yy))) - test_scores.set_index(["family", "step"]).loc[m, "test_roc_auc"])
              for m, (g, yy, _) in prep.items())
print(f"Largest gap between the NumPy AUC and scikit-learn (equal weights): {max_gap:.2e}")
assert max_gap < 1e-9, "The fast AUC does not match scikit-learn."
```

```python
rng = np.random.default_rng(RANDOM_STATE)
auc_draws = {m: [] for m in MODELS}
t0 = time.time()
for it in range(N_BOOT):
    counts = np.bincount(rng.integers(0, n_q, size=n_q), minlength=n_q).astype(float)
    if (counts[test_qidx] * test_y).sum() == 0 or (counts[test_qidx] * (1 - test_y)).sum() == 0:
        continue
    for m in MODELS:
        gid, yy, qs = prep[m]
        auc_draws[m].append(weighted_auc(gid, yy, counts[qs]))
auc_draws = {m: np.array(v) for m, v in auc_draws.items()}
n_used = len(auc_draws[MODELS[0]])
print(f"{n_used} of {N_BOOT} draws used, {time.time() - t0:.0f}s")

ts = test_scores.set_index(["family", "step"])
roc_ci = pd.DataFrame([{"family": m[0], "step": m[1], "test_roc_auc": ts.loc[m, "test_roc_auc"],
                        "ci_lo": float(np.percentile(auc_draws[m], 2.5)),
                        "ci_hi": float(np.percentile(auc_draws[m], 97.5))} for m in MODELS])
roc_ci.round(4)
```

```python
COMPARISONS = [("step2_minus_baseline", "step2_rate_level", "baseline"),
               ("step3_minus_step2", "step3_rate_change", "step2_rate_level"),
               ("step3_minus_baseline", "step3_rate_change", "baseline")]


def verdict(point, lo, hi):
    if lo > 0 and point >= MIN_GAIN:
        return "helps"
    if lo > 0:
        return "positive but below +0.01"
    if hi < 0:
        return "hurts"
    return "no significant effect"


rows = []
for fam in FAMILIES:
    for name, a, b in COMPARISONS:
        arr = auc_draws[(fam, a)] - auc_draws[(fam, b)]
        lo, hi = np.percentile(arr, [2.5, 97.5])
        point = float(ts.loc[(fam, a), "test_roc_auc"] - ts.loc[(fam, b), "test_roc_auc"])
        rows.append({"family": fam, "comparison": name, "point_delta": point, "ci_lo": float(lo),
                     "ci_hi": float(hi), "n_boot_used": len(arr), "verdict": verdict(point, lo, hi)})
test_deltas = pd.DataFrame(rows)
test_deltas.round(4)
```

```python
robust_rows = []
for name, _, _ in COMPARISONS:
    sub = test_deltas[test_deltas["comparison"] == name]
    n_help, n_hurt = int((sub["verdict"] == "helps").sum()), int((sub["verdict"] == "hurts").sum())
    if n_help >= 2:
        overall = "robust: helps in at least 2 of 3 families"
    elif n_help == 1:
        overall = "not robust: helps in only 1 family"
    elif n_hurt >= 1:
        overall = f"no gain; hurts in {n_hurt} family/families"
    else:
        overall = "no family shows a gain that clears the rule"
    robust_rows.append({"comparison": name, "families_that_help": n_help,
                        "families_that_hurt": n_hurt, "overall": overall})
pd.DataFrame(robust_rows)
```

```python
# Differences between families at each step, from the same draws.
FAMILY_PAIRS = [("random_forest", "logreg"), ("random_forest", "hist_gb"), ("hist_gb", "logreg")]
rows = []
for step in STEP_ORDER:
    for a, b in FAMILY_PAIRS:
        arr = auc_draws[(a, step)] - auc_draws[(b, step)]
        lo, hi = np.percentile(arr, [2.5, 97.5])
        point = float(ts.loc[(a, step), "test_roc_auc"] - ts.loc[(b, step), "test_roc_auc"])
        rows.append({"step": step, "comparison": f"{a} minus {b}", "point_delta": point,
                     "ci_lo": float(lo), "ci_hi": float(hi), "interval_excludes_zero": bool(lo > 0 or hi < 0)})
family_deltas = pd.DataFrame(rows)
family_deltas.round(4)
```

```python
# PR-AUC interval for the recommended model only (scikit-learn with quarter weights).
def weighted_ap(y, p, w):
    return average_precision_score(y, p, sample_weight=w)

unit_gap = abs(weighted_ap(test_y, test_p[REC], np.ones(len(test_y))) - ts.loc[REC, "test_pr_auc"])
assert unit_gap < 1e-9
rng2 = np.random.default_rng(RANDOM_STATE)
ap_draws = []
t0 = time.time()
for it in range(N_BOOT):
    counts = np.bincount(rng2.integers(0, n_q, size=n_q), minlength=n_q).astype(float)
    w = counts[test_qidx]
    if (w * test_y).sum() == 0:
        continue
    ap_draws.append(weighted_ap(test_y, test_p[REC], w))
ap_lo, ap_hi = np.percentile(ap_draws, [2.5, 97.5])
print(f"{REC[0]} / {REC[1]}: Test PR-AUC {ts.loc[REC, 'test_pr_auc']:.4f} "
      f"(95% interval {ap_lo:.4f} to {ap_hi:.4f}) against a base rate of {100 * base_rate_test:.2f}% "
      f"| ROC-AUC {ts.loc[REC, 'test_roc_auc']:.4f} (95% interval "
      f"{roc_ci.set_index(['family', 'step']).loc[REC, 'ci_lo']:.4f} to {roc_ci.set_index(['family', 'step']).loc[REC, 'ci_hi']:.4f}) "
      f"| {len(ap_draws)} draws, {time.time() - t0:.0f}s")
```

**Result on Test.** No step passes the rule (interval excludes zero and a gain of at least +0.01) in any family, so the rate adds nothing robust on Test either, including the 2022-23 hiking cycle. Adding the rate level lowers ROC-AUC for the random forest (-0.0077, interval -0.0118 to -0.0044) and histogram gradient boosting (-0.0223, -0.0414 to -0.0084) and leaves logistic regression unchanged (+0.0009, -0.0021 to +0.0035). Adding the rate change and interaction on top changes nothing significant in any family. The intervals are wide because Test has only 17 quarters. For logistic regression the interval of step 3 compared with step 2 reaches +0.0188, so on Test alone a gain of 0.01 cannot be ruled out there (for step 2 compared with the baseline the upper end is only +0.0035).

**Families.** At baseline, histogram gradient boosting is ahead of the random forest by 0.0058 (interval 0.0022 to 0.0107), and the random forest is ahead of logistic regression by 0.0189 (0.0008 to 0.0470). The random forest remains the recommended model, because it was chosen on validation, where the two were 0.0014 apart. This difference is reported and is not used to change the choice. The PR-AUC of the recommended model is 0.4665, with a 95% interval of 0.3379 to 0.5738, against a base rate of 8.11%.

## 9. Calibration

Calibration asks whether a predicted probability of 20% really means that about 20% of such counties decline. Test is split into probability deciles and compared with the same check on the validation predictions. No recalibration is applied: the check shows whether one is needed.

```python
def calib_table(y, p):
    d = pd.DataFrame({"p": p, "y": y})
    d["decile"] = pd.qcut(d["p"], 10, duplicates="drop")
    t = d.groupby("decile", observed=True).agg(mean_predicted=("p", "mean"), actual_rate=("y", "mean"), n=("y", "size"))
    return t.reset_index(drop=True)


cal_val = calib_table(val_y, val_p[REC]); cal_val.insert(0, "split", "validation")
cal_test = calib_table(test_y, test_p[REC]); cal_test.insert(0, "split", "test")
cal_rec = pd.concat([cal_val, cal_test]).reset_index().rename(columns={"index": "decile"})
cal_rec["decile"] = cal_rec["decile"] + 1
print(f"Recommended model ({REC[0]}, {REC[1]}):")
print(f"  validation: mean predicted {100 * val_p[REC].mean():.2f}% vs actual {100 * val_y.mean():.2f}%, "
      f"Brier {brier_score_loss(val_y, val_p[REC]):.4f}")
print(f"  test:       mean predicted {100 * test_p[REC].mean():.2f}% vs actual {100 * test_y.mean():.2f}%, "
      f"Brier {brier_score_loss(test_y, test_p[REC]):.4f}")
cal_rec.round(4)
```

```python
rows = []
for m in MODELS:
    ct, cv = calib_table(test_y, test_p[m]), calib_table(val_y, val_p[m])
    rows.append({"family": m[0], "step": m[1],
                 "val_mean_pred_pct": 100 * val_p[m].mean(), "val_actual_pct": 100 * val_y.mean(),
                 "test_mean_pred_pct": 100 * test_p[m].mean(), "test_actual_pct": 100 * test_y.mean(),
                 "test_brier": brier_score_loss(test_y, test_p[m]),
                 "test_deciles_underconfident": int((ct["actual_rate"] > ct["mean_predicted"]).sum()),
                 "test_deciles": len(ct)})
calib_all = pd.DataFrame(rows)
calib_all.round(4)
```

**Reading the calibration.** For the recommended model the average predicted probability is 6.91% on Test against an actual 8.11%, and it is below the actual rate in all 10 deciles (for example 1.3% predicted against 2.9% actual in decile 6). The gap is smallest in the riskiest tenth (46.1% predicted against 47.6% actual) and largest in deciles 5 to 8, where the risk is understated by a factor of about 1.5 to 3. On validation the direction is the opposite: 5.39% predicted against 3.77% actual, with the top decile at 38.1% predicted against 25.5% actual. A possible reason is that the models carry over the decline rate of the years they were trained on: in 9 of the 11 folds (see the fold table of Phase 5) the validation year had a lower rate than its training years, and Test has a higher rate than Train (8.11% against 4.85%). This was not tested separately. The consequence is that the ranking is useful, but a predicted probability is not a stable estimate of the share that will decline. A recalibration fitted on validation would move the probabilities in the wrong direction for Test, so none is applied. Adding the rate level raises the average predicted probability (8.48% for logistic regression, 9.65% for the random forest), but it does not raise the ROC-AUC.

## 10. Stability over time: each Test quarter

The Test period has 17 quarters with very different decline rates. For every quarter this shows the base rate, the ROC-AUC, the PR-AUC and the recall at the riskiest 10% of that quarter, for the recommended model, and then the ROC-AUC of all nine models side by side. A quarter without both outcomes has no AUC.

```python
def quarter_metrics(m):
    out = []
    for q in test_quarters:
        mask = (test["quarter"] == q).to_numpy()
        yq, pq = test_y[mask], test_p[m][mask]
        both = 0 < yq.sum() < len(yq)
        k = max(1, int(round(0.10 * len(yq))))
        flag = np.zeros(len(yq), dtype=bool); flag[np.argsort(-pq, kind="mergesort")[:k]] = True
        out.append({"quarter": pd.Timestamp(q).date(), "family": m[0], "step": m[1], "n": int(mask.sum()),
                    "positives": int(yq.sum()), "base_rate_pct": 100 * yq.mean(),
                    "roc_auc": roc_auc_score(yq, pq) if both else np.nan,
                    "pr_auc": average_precision_score(yq, pq) if yq.sum() > 0 else np.nan,
                    "top10_recall": (flag & (yq == 1)).sum() / yq.sum() if yq.sum() > 0 else np.nan,
                    "top10_precision": (flag & (yq == 1)).sum() / flag.sum()})
    return out


per_quarter = pd.DataFrame([r for m in MODELS for r in quarter_metrics(m)])
rec_q = per_quarter[(per_quarter["family"] == REC[0]) & (per_quarter["step"] == REC[1])]
print(f"Recommended model ({REC[0]}, {REC[1]}) in each Test quarter:")
rec_q.drop(columns=["family", "step"]).round(4).reset_index(drop=True)
```

## 12. The fixed step order (addition from the team merge)

The changes measured above depend on the order in which the features were added: momentum first, then the rate level, then the rate change and the interaction. They measure what the mortgage rate adds **once a county's own price momentum is already known**, not the importance of the rate on its own. A different order (rate first, momentum added on top) would be expected to give the rate a larger share, because momentum would no longer be known when the rate's contribution is measured. That order was not run, and the results must not be read as if it had been.
